```python
import requests
import json
import time
print("libraries loaded")
```

```python
# pulling the full load - pages 0 to 19, thats around 20,000 games
# steamspy rate limits us so we wait 1.5 seconds between each page or it blocks us

all_games = {}

for page in range(20):
    url = f"https://steamspy.com/api.php?request=all&page={page}"
    response = requests.get(url)
    data = response.json()
    all_games.update(data)
    print(f"page {page} done - total games collected: {len(all_games)}")
    time.sleep(1.5)

print("full load done!")
```

```python
# saving the full load as a json file
with open("bronze_full_load_sample.json", "w") as f:
    json.dump(all_games, f)

print("file saved successfully")
```

# **Daily Incremental Load**

```python
# pulling page 0 only - this is what our daily incremental load looks like
# page 0 has the most active games and gets refreshed by steamspy every day

incremental_data = {}

url = "https://steamspy.com/api.php?request=all&page=0"
response = requests.get(url)
incremental_data = response.json()

# tagging each record with todays date so we know when it was fetched
from datetime import date
today = str(date.today())

for game in incremental_data.values():
    game["date_fetched"] = today

print(f"incremental load done - {len(incremental_data)} games fetched for {today}")
```

```python
# saving the incremental load file
with open("bronze_incremental_load_sample.json", "w") as f:
    json.dump(incremental_data, f)

print("incremental file saved successfully")
```

# **Fetching steamstore data**

```python
import time
from datetime import date

today = str(date.today())
incremental_store = {}
failed = []

# steam store api only allows 1 appid at a time on the free endpoint
# we batch them in groups of 10 with a small wait to avoid getting blocked

for i, appid in enumerate(top_appids):
    try:
        url = f"https://store.steampowered.com/api/appdetails?appids={appid}"
        response = requests.get(url, timeout=10)
        data = response.json()
        
        if data and data.get(appid, {}).get('success'):
            game_data = data[appid]['data']
            incremental_store[appid] = {
                'appid': int(appid),
                'name': game_data.get('name'),
                'is_free': game_data.get('is_free'),
                'recommendations': game_data.get('recommendations', {}).get('total', 0),
                'genres': [g['description'] for g in game_data.get('genres', [])],
                'categories': [c['description'] for c in game_data.get('categories', [])],
                'release_date': game_data.get('release_date', {}).get('date'),
                'date_fetched': today
            }
        else:
            failed.append(appid)
            
    except Exception as e:
        failed.append(appid)
    
    # print progress every 50 games
    if (i + 1) % 50 == 0:
        print(f"done {i+1}/1000 - successful: {len(incremental_store)} - failed: {len(failed)}")
    
    time.sleep(1.5)

print(f"\nfinished - total fetched: {len(incremental_store)} - failed: {len(failed)}")
```

```python
with open("steam_store_incremental_2026-10-09.json", "w") as f:
    json.dump(incremental_store, f)

print("saved successfully")
```
